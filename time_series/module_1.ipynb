```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
ts = pd.Series([1.1, 1.3, None, 1.2, 1.4, None, 1.3],
               index=pd.date_range(start='2024-01-01', periods=7, freq='D'))
# заполнение средним значением
ts_filled_mean = ts.fillna(ts.mean())
# заполнение медианой
ts_filled_median = ts.fillna(ts.median())
```

```python
ts_missing = ts.copy()
ts_missing.iloc[5:10] = np.nan  # удалим 5 подряд идущих значений
ts_filled = ts_missing.ffill()
ts_interpolated = ts_missing.interpolate()
# Строим графики
ts_missing.plot(label='С пропусками', style='--', color='gray')
ts_filled.plot(label='Forward fill', color='blue')
ts_interpolated.plot(label='Интерполяция', color='green')
plt.title('Сравнение методов заполнения пропусков')
plt.legend()
plt.grid(True)
plt.show()
```
